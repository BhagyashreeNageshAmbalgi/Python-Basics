Q14) Describe a scenario where a "while" loop is more suitable than a "for" loop?

Ans: "for" loops are used best when number of iterations are predefined but when in such condition where we dont know number of iterations in advance then we "while" is best used.




```python
import random

secret_number = random.randint(1, 10)
attempts = 0

while True:
    user_guess = int(input("Guess a number between 1 and 10: "))
    attempts += 1
    if user_guess == secret_number:
        print(f"Congratulations! You found the number in {attempts} attempts.")
        break
    elif user_guess < secret_number:
        print("Too low! Try again.")
    else:
        print("Too high! Try again.")
```

in this above example,  a while loop is more suitable than a for loop because it allows for conditional repetition and handles the unkown number of iterations.

Q4) Write a program that checks if a given word is python keyword



```python
import keyword

def is_python_keyword(word):
    return keyword.iskeyword(word)

def main():
    word = input("Enter a word: ")
    if is_python_keyword(word):
        print(f"'{word}' is a Python keyword.")
    else:
        print(f"'{word}' is not a Python keyword.")

if __name__ == "__main__":
    main()
```

 Q5) Create a list and tuple in Python, and demonstrate how attempting to change an element works differently
for each.

```python
# Create a list and a tuple
my_list = [1, 2, 3, 4, 5]
my_tuple = (1, 2, 3, 4, 5)

# Print the original values
print("Original List:", my_list)
print("Original Tuple:", my_tuple)

# Attempt to change an element in the list
my_list[0] = 10
print("List after modification:", my_list)

# Attempt to change an element in the tuple
try:
    my_tuple[0] = 10
except TypeError as e:
    print("Error:", e)
    print("Tuples are immutable and cannot be changed after creation.")
```

6. Write a function to demonstrate the behavior of mutable and immutable arguments.

```python
def modify_arguments(my_list, my_tuple, my_string, my_int):
    # Modify the list
    my_list.append(10)
    print("List inside function:", my_list)

    # Attempt to modify the tuple
    try:
        my_tuple[0] = 10
    except TypeError as e:
        print("Error:", e)

    # Modify the string
    my_string += " World"
    print("String inside function:", my_string)

    # Modify the integer
    my_int += 10
    print("Integer inside function:", my_int)

# Create a list, tuple, string, and integer
my_list = [1, 2, 3]
my_tuple = (1, 2, 3)
my_string = "Hello"
my_int = 10

# Print the original values
print("Original List:", my_list)
print("Original Tuple:", my_tuple)
print("Original String:", my_string)
print("Original Integer:", my_int)

# Call the function
modify_arguments(my_list, my_tuple, my_string, my_int)

# Print the values after function call
print("List after function call:", my_list)
print("Tuple after function call:", my_tuple)
print("String after function call:", my_string)
print("Integer after function call:", my_int)
```

Q7)  Write a program that performs basic arithmetic operations on two user-input numbers.

```python
def get_user_input():
    try:
        num1 = float(input("Enter the first number: "))
        num2 = float(input("Enter the second number: "))
        return num1, num2
    except ValueError:
        print("Invalid input. Please enter a number.")
        return get_user_input()

def perform_arithmetic_operations(num1, num2):
    print("Arithmetic Operations:")
    print(f"Addition: {num1} + {num2} = {num1 + num2}")
    print(f"Subtraction: {num1} - {num2} = {num1 - num2}")
    print(f"Multiplication: {num1} * {num2} = {num1 * num2}")
    if num2 != 0:
        print(f"Division: {num1} / {num2} = {num1 / num2}")
    else:
        print("Error: Division by zero is not allowed.")

def main():
    num1, num2 = get_user_input()
    perform_arithmetic_operations(num1, num2)

if __name__ == "__main__":
    main()
```

 Q8) Write a program to demonstrate the use of logical operators.

```python
def demonstrate_logical_operators():
    # Define variables
    x = 5
    y = 10

    # Logical AND (and)
    print("Logical AND (and):")
    print(f"x > 0 and y > 0: {x > 0 and y > 0}")  # True
    print(f"x > 0 and y < 0: {x > 0 and y < 0}")  # False

    # Logical OR (or)
    print("\nLogical OR (or):")
    print(f"x > 0 or y > 0: {x > 0 or y > 0}")  # True
    print(f"x < 0 or y < 0: {x < 0 or y < 0}")  # False

    # Logical NOT (not)
    print("\nLogical NOT (not):")
    print(f"not (x > 0): {not (x > 0)}")  # False
    print(f"not (x < 0): {not (x < 0)}")  # True

def main():
    demonstrate_logical_operators()

if __name__ == "__main__":
    main()
```

 Q9) Write a Python program to convert user input from string to integer, float, and boolean types

```python
def convert_types():
    user_input = input("Enter a value: ")

    try:
        # Convert to integer
        int_value = int(user_input)
        print(f"Integer value: {int_value}")
    except ValueError:
        print("Cannot convert to integer.")

    try:
        # Convert to float
        float_value = float(user_input)
        print(f"Float value: {float_value}")
    except ValueError:
        print("Cannot convert to float.")

    # Convert to boolean
    if user_input.lower() == "true":
        bool_value = True
    elif user_input.lower() == "false":
        bool_value = False
    else:
        bool_value = bool(user_input)
        print("Note: Non-empty strings are considered True, empty strings are considered False.")
    print(f"Boolean value: {bool_value}")

def main():
    convert_types()

if __name__ == "__main__":
    main()
```

 Q10) Write code to demonstrate type casting with list elements.

```python
def type_casting_list():
    # Create a list of strings
    string_list = ["1", "2", "3", "4", "5"]
    print("Original List:", string_list)

    # Convert string list to integer list
    try:
        int_list = [int(x) for x in string_list]
        print("Integer List:", int_list)
    except ValueError:
        print("Error: Cannot convert all elements to integer.")

    # Convert string list to float list
    try:
        float_list = [float(x) for x in string_list]
        print("Float List:", float_list)
    except ValueError:
        print("Error: Cannot convert all elements to float.")

def main():
    type_casting_list()

if __name__ == "__main__":
    main()
```

Q11) Write a program that checks if a number is positive, negative, or zero.

 Q13) Write a Python program to find the sum of all even numbers between 1 and 50

```python
def sum_even_numbers():
    total = 0
    for i in range(1, 51):
        if i % 2 == 0:
            total += i
    return total

def main():
    result = sum_even_numbers()
    print(f"The sum of even numbers between 1 and 50 is: {result}")

if __name__ == "__main__":
    main()
```

Q14) Write a program to reverse a string using a while loop.

```python
def reverse_string(s):
    reversed_s = ""
    i = len(s) - 1
    while i >= 0:
        reversed_s += s[i]
        i -= 1
    return reversed_s

def main():
    s = input("Enter a string: ")
    reversed_s = reverse_string(s)
    print(f"Reversed string: {reversed_s}")

if __name__ == "__main__":
    main()
```

Q15) Write a Python program to calculate the factorial of a number provided by the user using a while loop

```python
def calculate_factorial(n):
    factorial = 1
    i = 1
    while i <= n:
        factorial *= i
        i += 1
    return factorial

def main():
    try:
        n = int(input("Enter a non-negative integer: "))
        if n < 0:
            print("Error: Factorial is not defined for negative numbers.")
        else:
            result = calculate_factorial(n)
            print(f"The factorial of {n} is: {result}")
    except ValueError:
        print("Invalid input. Please enter an integer.")

if __name__ == "__main__":
    main()
```
